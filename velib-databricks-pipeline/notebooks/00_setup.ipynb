# 00 - Setup du projet Vélib'
**Objectif :** préparer la structure de stockage du projet.

Architecture médaillon :
- **Bronze** : données brutes, telles qu'elles arrivent
- **Silver** : données nettoyées
- **Gold** : agrégats prêts pour le dashboard

%md
## 1. Création des schémas et du volume
On crée 3 schémas dans le catalogue `workspace` (un par couche), plus un volume `raw` dans `velib_bronze` pour déposer les fichiers JSON de Vélib'.

```sql
%sql
-- Les 3 couches du projet
CREATE SCHEMA IF NOT EXISTS workspace.velib_bronze;
CREATE SCHEMA IF NOT EXISTS workspace.velib_silver;
CREATE SCHEMA IF NOT EXISTS workspace.velib_gold;

-- L'endroit où déposer les fichiers bruts
CREATE VOLUME IF NOT EXISTS workspace.velib_bronze.raw;
```

%md
## 2. Vérification
On lit les fichiers déposés dans le volume pour vérifier que tout fonctionne.

```sql
%python
dossier = "/Volumes/workspace/velib_bronze/raw"

# Infos fixes des stations : nom, position GPS, capacité
df_info = spark.read.json(f"{dossier}/station_information_2026-10-24_1848.json")
display(df_info)

# Disponibilité à un instant T : vélos dispos, places libres
df_status = spark.read.json(f"{dossier}/station_status_2026-10-24_1852.json")
display(df_status)
```
